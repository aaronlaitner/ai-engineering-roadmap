# 1. Confusion Matrix

For a binary classifier, there are four possible outcomes:

|  | Actually Positive | Actually Negative |
|---|---:|---:|
| **Predicted Positive** | True Positive (TP) | False Positive (FP) |
| **Predicted Negative** | False Negative (FN) | True Negative (TN) |

A useful way to remember the terms:

- **Positive / Negative** describes the model's prediction.
- **True / False** describes whether that prediction was correct.

So:

- **TP** = predicted positive and was correct
- **FP** = predicted positive and was wrong
- **TN** = predicted negative and was correct
- **FN** = predicted negative and was wrong


# 2. Accuracy

Accuracy asks:

> What fraction of all predictions were correct?

$$
\text{Accuracy}
=
\frac{TP+TN}{TP+TN+FP+FN}
$$

Example:

$$
TP=32,\ FN=8,\ FP=12,\ TN=148
$$

Then:

$$
\text{Accuracy}
=
\frac{32+148}{200}
=
0.90
$$

Accuracy can be misleading on highly imbalanced datasets. If only 1% of examples are positive, a classifier that predicts every example as negative would still achieve 99% accuracy.


# 3. Precision

Precision asks:

> When the model predicts positive, how often is it correct?

$$
\text{Precision}
=
\frac{TP}{TP+FP}
$$

The denominator contains **all predicted positives**.

Using $TP=32$ and $FP=12$:

$$
\text{Precision}
=
\frac{32}{44}
\approx
0.727
$$

Precision matters especially when **false positives are costly**.


# 4. Recall / True Positive Rate

Recall asks:

> Of all the actual positives, how many did the model successfully detect?

$$
\text{Recall}
=
\frac{TP}{TP+FN}
$$

Recall is also the **True Positive Rate (TPR)**.

Using $TP=32$ and $FN=8$:

$$
\text{Recall}
=
\frac{32}{40}
=
0.80
$$

Recall matters especially when **false negatives are costly**.


# 5. Specificity and False Positive Rate

Specificity asks:

> Of all the actual negatives, how many did the model correctly classify as negative?

$$
\text{Specificity}
=
\frac{TN}{TN+FP}
$$

The False Positive Rate asks:

> Of all actual negatives, how many were incorrectly classified as positive?

$$
\text{FPR}
=
\frac{FP}{FP+TN}
$$

and:

$$
\boxed{\text{FPR}=1-\text{Specificity}}
$$

Important:

**FPR is not the opposite of recall.**

- Recall uses actual positives.
- FPR uses actual negatives.


# 6. F1 Score

F1 combines precision and recall:

$$
F1
=
2\frac{\text{Precision}\cdot\text{Recall}}
{\text{Precision}+\text{Recall}}
$$

It uses the harmonic mean, so it penalizes situations where one of precision or recall is high while the other is poor.

Using precision $\approx0.727$ and recall $=0.80$:

$$
F1\approx0.762
$$


# 7. Classification Thresholds

Many classifiers output a score rather than an immediate class label.

Example:

| Example | Positive-class score |
|---|---:|
| A | 0.97 |
| B | 0.81 |
| C | 0.61 |
| D | 0.48 |
| E | 0.20 |

At threshold $0.50$, A, B, and C are positive.

At threshold $0.30$, A, B, C, and D are positive.

On a fixed dataset with fixed model scores, lowering the threshold means:

- TP can stay the same or increase
- FP can stay the same or increase
- FN can stay the same or decrease
- TN can stay the same or decrease
- recall can stay the same or increase

Precision often decreases, but not necessarily at every individual threshold step.


# 8. ROC Curves

A ROC curve shows classifier behavior across many thresholds.

Its axes are:

$$
\text{x-axis}=\text{False Positive Rate}
$$

$$
\text{y-axis}=\text{True Positive Rate}=\text{Recall}
$$

where:

$$
\text{TPR}=\frac{TP}{TP+FN}
$$

and:

$$
\text{FPR}=\frac{FP}{FP+TN}
$$

A point on the ROC curve is therefore:

$$
(\text{FPR},\text{TPR})
$$

A strong classifier bends toward the **top-left**, where recall is high and false-positive rate is low.


# 9. ROC-AUC

ROC-AUC means **Area Under the ROC Curve**.

Useful reference points:

- $0.5$ → roughly random ranking
- $1.0$ → perfect ranking

A useful interpretation is:

> ROC-AUC is the probability that a randomly chosen positive example receives a higher score than a randomly chosen negative example.

So:

$$
\text{ROC-AUC}=0.92
$$

means a random positive is ranked above a random negative about **92% of the time**.

It does **not** mean that 92% of positive predictions are correct.


# 10. Precision-Recall Curves

A Precision-Recall curve varies the classification threshold while plotting:

$$
\text{x-axis}=\text{Recall}
$$

$$
\text{y-axis}=\text{Precision}
$$

As the threshold is lowered:

- recall usually increases
- precision often decreases

PR curves are especially useful when the positive class is rare.


# 11. Why Class Imbalance Matters

Suppose there are:

$$
100\text{ actual positives}
$$

and:

$$
9900\text{ actual negatives}
$$

and a model produces:

$$
TP=90,\quad FN=10,\quad FP=900,\quad TN=9000
$$

Then:

$$
\text{Recall}
=
\frac{90}{100}
=
0.90
$$

and:

$$
\text{FPR}
=
\frac{900}{9900}
\approx
0.091
$$

but:

$$
\text{Precision}
=
\frac{90}{90+900}
=
\frac{1}{11}
\approx
0.091
$$

Even a modest false-positive rate can create many false positives when the negative class is enormous.


# 12. ROC-AUC vs PR-AUC

## ROC-AUC

$$
\boxed{\text{How well does the model rank/separate positives from negatives overall?}}
$$

ROC uses:

- Recall / TPR
- False Positive Rate

## PR-AUC

$$
\boxed{\text{How well does the model maintain precision while finding positives?}}
$$

PR uses:

- Precision
- Recall

PR-AUC is often more revealing when:

- the positive class is rare
- false positives matter
- class imbalance is substantial

For a random classifier, the PR baseline depends on positive-class prevalence.


# 13. NumPy Example

Calculate the major metrics directly from a confusion matrix.


```python
TP = 32
FN = 8
FP = 12
TN = 148

accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
specificity = TN / (TN + FP)
fpr = FP / (FP + TN)
f1 = 2 * (precision * recall) / (precision + recall)

print(f"Accuracy:    {accuracy:.3f}")
print(f"Precision:   {precision:.3f}")
print(f"Recall/TPR:  {recall:.3f}")
print(f"Specificity: {specificity:.3f}")
print(f"FPR:         {fpr:.3f}")
print(f"F1:          {f1:.3f}")
```

# 14. Threshold Example

This demonstrates how changing a threshold changes labels.


```python
import numpy as np

scores = np.array([0.97, 0.81, 0.61, 0.48, 0.20])

for threshold in [0.50, 0.30]:
    predictions = (scores >= threshold).astype(int)
    print(f"Threshold {threshold:.2f}: {predictions}")
```

# 15. Final Summary

| Metric | Main Question |
|---|---|
| Accuracy | How often is the model correct overall? |
| Precision | When the model predicts positive, how often is it correct? |
| Recall / TPR | Of all actual positives, how many did the model find? |
| Specificity | Of all actual negatives, how many did the model correctly reject? |
| FPR | Of all actual negatives, how many did the model falsely flag? |
| F1 | How well are precision and recall balanced? |
| ROC-AUC | How well does the model rank positives above negatives across thresholds? |
| PR-AUC | How well does the model maintain precision while recovering positives? |

Threshold intuition:

$$
\text{lower threshold}
\Rightarrow
\text{more positive predictions}
\Rightarrow
\text{higher recall, usually lower precision}
$$

Class imbalance intuition:

> A small false-positive rate can still create many false positives if the negative class is very large.
