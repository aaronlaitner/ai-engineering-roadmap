Constructors are functions called when an object is instantiated.

__init__() function is used as the constructor and is called automatically when an object is created.

self parameter refers to the current instance

```python
class ClassSchedule:
  # Self defines the variables associated with the instance of the object. Aka variables with self before it are variables
  # that can be accessed with objectinstancename.variablename

  # init function is called upon object creation
  def __init__(self, course):
    self.course = course

# Creates object and assigns name Chemistry to the instance variable course
first = ClassSchedule('Chemistry')
#objectname.instancevariable allows us to access the var value inside the instance variable belonging to the object
print(first.course)
```

Destructors are special functions that are called when an object gets deleted. In Python, the __del__() method is commonly used as the destructor and is called when an object is deleted.



```python
class ClassSchedule:
   def __init__(self, course):
       self.course = course

   def __del__(self):
       print('You successfully deleted your schedule')

# Eg.) of deconstructor being used
newSched = ClassSchedule("Chemistry")
# To remove the object use del, this triggers the delete function inside the object inherited from the class
del newSched
```

Members are known as the methods and variables inside a class. Access modifiers are used to control or restrict access to members.

There are 3 types of modifiers in Python:
- Public Access Modifiers
- Protected Access Modifiers
- Private Access Modifiers

By default, all members within a class are public. Being public means all members can be accessed outside of the class in another part of the program.

```python
# All members part of the class are public
class ClassSchedule:
   def __init__(self, course, instructor):
       self.course = course
       self.instructor = instructor

   def display_course(self):
       print(f'Course: {self.course}, Instructor: {self.instructor}')

sched = ClassSchedule('Chemistry', 'Mr. Doe') # initializing

# We can run the methods and variables and access them because the members are public
sched.display_course() # prints Course: Chemistry, Instructor: Mr. Doe
sched.course # prints 'Chemistry

```

Protected Access Modifiers are denoted with prefix __ (2 of these _!) and declare the members to only be accessible within the class only. Any attempts to utilize them/access them outside the class will cause an Attribute Error Message

```python
# Can have a mixture of public and private members
class PrivateDog:
  def __init__(self, name):
    self.__name = name
  def display_name(self):
    print(f'Dog name is {self.__name}')

Sparky = PrivateDog('John')
# Trynna print dog name will cause error as we're attempting to access a private member
Sparky.__name
```

```python
# However the method is not private!
Sparky.display_name()
```

Encapsulation describes wrapping variables and methods in one unit. A class is a popular example of encapsulation, as it "encapsualtes" members such as variables and methods into a single unit.

Inheritance is the ability that enables the transfer of methods and properties of one class to another.

There are two properties to inheritance:
- Parent class
- Child class

Parent class (aka base class) is the class whose methods and properties transfer over to the child class

```python
# Example of parent class
class Person:
   def __init__(self, name, age):
       self.name = name
       self.age = age

   def print_info(self):
       print(f'{self.name} is {self.age} years old')
```

Child class (aka derived class) is the class that inherits methods and properties from the parent class. Child class must contain the following:
- Name of parent class in the definition of the child class
- Constructor of the parent class called within the constructor of the child class

Following example shows Person as the parent class and the child class named Teacher

```python
# Parent class
class Person:
  def __init__(self, name, age):
      self.name = name
      self.age = age
  def print_info(self):
      print(self.name)
      print(self.age)
# Child class
class Teacher(Person): # note ref to parent class here
  def __init__(self, name, age, subject):
      self.subject = subject

      # Can also use super().__init__(name, age)
      Person.__init__(self, name, age) # note these must be a subset of vars of parent init func! (aka must be name, age or subject)
```

Polymorphism is the concept of classes sharing methods with the same name. This becomes useful when there are objects initiated from different classes that may share similar methods.

Polymorphism with classes simply means classes are allowed to contain methods which share the same name as methods from another class

```python
# Polymorphism in OOP is the concept of classes sharing methods with the same name.
class Checking():
   def type(self):
       print('You have a checking account at the Codecademy Bank.')

   def balance(self):
       print('$20 left in your checking.')

class Savings():
   def type(self):
       print('You have a savings account at the Codecademy Bank.')

   def balance(self):
       print('$1000 left in your savings.')

# Polymorhism is helpful as we can reduce need for excess code
account_a = Checking()
account_b = Savings()

for account in (account_a, account_b):
   account.type()
   account.balance()
```

__str__ is a special method which automatically is called when an instance calls the str() method. Print when used on an object makes the instance call the str() method. As such the main point is that when you call print on a normal object it will print the memory address of the instance, the __str__ method allows you to customize what happens when print is used on an object, such as printing out the values of the variables of that object. (Pulled from https://www.pythontutorial.net/python-oop/python-__str__/)

```python
class Person:
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age

    def __str__(self):
        return f'Person({self.first_name},{self.last_name},{self.age})'

# Eg str converts other data types of a variable to a string
n = 123
s = str(n) # converts datatype to "123"
print(s)

# Here's how it works for class
person1 = Person("John", "Smith", "25")
# __str__ will run
str(person1)
# __str__ will run
print(person1)
```
