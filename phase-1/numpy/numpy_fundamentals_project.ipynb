```python
# Part 1: Create the dataset
# Use numpy to randomly generate a (100, 5) array containing integer grades from
#  40-100 then print shape, number of dimensions, and datatype

# Used notes: random
import numpy as np
# grades_array = np.random.randint(1, 10, size=(2, 3))
np.random.seed(42)
grades_array = np.random.randint(40, 101, size=(100, 5)) # IMPORTANT: 101 NOT 100!!

print(grades_array)
print(grades_array.shape) # Print shape
print(grades_array.ndim) # Print number dimensions
print(grades_array.dtype) # print data type

```

```python
# part 2: Basic statistics

# No notes used

# Calculate avg grade across the entire dataset
print(grades_array.mean())

# print highest grade
print(grades_array.max())

# print lowest grade
print(grades_array.min())

# print avg grade for each assessment (column)
print(grades_array.mean(axis=0))

# print avg grade for each student (rows)
print(grades_array.mean(axis=1))

# Print std for each assessment
print(grades_array.std(axis=0))
```

```python
# Part 3: Indexing and slicing

# No notes were used

# Extract all grades belonging to the first 10 students
# Aka get first 10 rows of data (slicing)
print(f'first 10 rows:\n {grades_array[:10]}')

# Extract grades for assessment #3 only
print(f'get grades for 3rd assessment:\n {grades_array[:,2]}')

# Extract grades for first 10 students on the first 3 assessments
# Get first 10 rows of data, and first 3 columns
print(f'Get grades for first 10 students on the first 3 assessments:\n {grades_array[:10,:3]}')

# Get the grade belonging to student # 25 on assessment #4
print(f'Get the grade belonging to student # 25 on assessment #4:\n {grades_array[24,3]}')

```

```python
# Part 4: Boolean masking

# Used no notes

# Find every individual grade >= 90
# uses boolean masking
print(f'Every individual grade over 90:\n {grades_array[grades_array >= 90]}')

# Find how many individual grades are >= 90
# uses boolean masking + len()
print(f'The number of grades over 90: {len(grades_array[grades_array >= 90])}')

# Find all students whose avg grades are >= 80 (should contain complete rows of grades + avgs)
l1 = grades_array[grades_array.mean(axis=1) >= 80] # avg grades above 80
# average across columns (assessments) producing one avg per student
print(f'All students scores whose avg grades are >= 80:\n {l1}')

# Find all students whose avg grade is < 60
l2 = grades_array[grades_array.mean(axis=1) < 60] # avg grades below 60
print(f'All students scores whose avg grades are < 60:\n {l2}')
```

```python
# Part 5: Conditional Replacement

# No notes were used

# Create a new array where:
# - grades < 50 become 0
# - everything else remains unchanged
# Use np.where
drop_scores_array = np.where(grades_array < 50, 0, grades_array)

print(f'New array:\n {drop_scores_array}')

# Verify that original array remains unchanged
print(f'Original data:\n {grades_array}')
```

```python
# Part 6: Give assessments different weights

# No notes were used to complete question

# The assessments are weighted:
weights = np.array([0.10, 0.15, 0.20, 0.25, 0.30])

# calculate the weighted final grade of every student without a python loop
# End result should have shape: (100,)

# Multiply each row by the weights and add them together (dot product!?)
weighted_array = grades_array @ weights

print(f"Verifying shape is correct: {weighted_array.shape}")

print(f"Final grades: {weighted_array}")
```

```python
# Part 7: Standarizing the dataset

# No notes used

# standardize each assessment independently so that each column (axis 0) approximately has:
# - mean = 0
# - standard deviation 1
# z = (x - mean) / std

mean = grades_array.mean(axis = 0)
stn_dev = grades_array.std(axis = 0)

standard_array_part1 = (grades_array - mean) / stn_dev

# Check that the standardization worked such that each column has mean 0 and std 1
print(f'Means:\n {standard_array_part1.mean(axis = 0)}') # Floating point close to 0
print(f'Standard Deviations:\n {standard_array_part1.std(axis = 0)}')
```

```python
# Part 8: Vectorization Challenge

# No notes used

# Now imagine the school decides to give everyone 5 bonus points, but grades cannot exceed 100
grade_inflation_raw = grades_array + 5

# np.where() can be used to replace any grades above 100 with 100 if need be
grade_inflation = np.where(grade_inflation_raw > 100, 100, grade_inflation_raw)

print(f'grades before 5 bonus points:\n {grades_array}')
print(f'grades after 5 bonus points:\n {grade_inflation}')

# Checking if any grades are above 100 (Should be an empty array)
print(f'Grades higher than 100:\n {grade_inflation[grade_inflation > 100]}')
```

```python
# Part 9: Reshaping

# used notes for reshape syntax

# Bonus: figure out how -1 could be used in one of those reshapes
# Take the first 20 students and reshape their data from (20, 5) into (10, 10)
reshape_array = grades_array[:20]
reshape_array = np.reshape(reshape_array, (10, 10))
print(f'Compact (10, 10):\n{reshape_array}')
# Then reshape it back into (20, 5)
reshape_array = np.reshape(reshape_array, (20, -1))
print(f'Back to original shape:\n{reshape_array}')

```

```python
# Part 10: Final challenge no loop summary

# No notes were used

print("NUMPY STUDENT PERFORMANCE REPORT\n")
print(f'# of students: {len(grades_array)}')
print(f'Assessments: {len(grades_array[0])}\n')

print(f'Overall mean: {grades_array.mean()}')
print(f'Highest grade: {grades_array.max()}')
print(f'Lowest grade: {grades_array.min()}\n')

print(f'Assessment means:\n{grades_array.mean(axis=0)}\n')
print(f'# of students with average >= 80: {len(grades_array[grades_array.mean(axis = 1) >= 80])}')
print(f'# of students with average < 60: {len(grades_array[grades_array.mean(axis = 1) < 60])}\n')

print(f'Highest weighted final grade: {weighted_array.max()}')
print(f'Lowest weighted final grade: {weighted_array.min()}')
```
